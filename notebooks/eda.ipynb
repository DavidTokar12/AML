# EDA

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

from scipy import stats

from aml.data import load_data


X_train, y_train, X_test = load_data()
print(X_train.shape, y_train.shape, X_test.shape)
```

```python
missing = X_train.isna().mean()
print(f"Missing: {missing.mean():.1%} overall, max per feature {missing.max():.1%}")
```

```python
cardinality = X_train.nunique().sort_values()
print("Constant features:", cardinality[cardinality <= 1].index.tolist())
sns.histplot(cardinality, bins=100)
plt.title("Feature cardinality")
plt.show()
```

```python
sns.heatmap(X_train.corr(), cmap="coolwarm", center=0)
plt.title("Feature correlation")
plt.show()
```

```python
z = np.abs(stats.zscore(X_train, nan_policy="omit"))
print("Samples with any |z| > 3:", int((z > 3).any(axis=1).sum()))
```

```python
sns.histplot(y_train, kde=True)
plt.title("Target (age)")
plt.show()
print(f"Skew {stats.skew(y_train):.3f}, kurtosis {stats.kurtosis(y_train):.3f}")
```
